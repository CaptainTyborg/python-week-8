```python
import pandas as pd
import cord19_analysis as ca

# Try to load a sample of metadata.csv (adjust nrows if needed)
try:
    df = ca.load_data('metadata.csv', nrows=20000)
    print('Loaded', df.shape)
except FileNotFoundError:
    print('Place metadata.csv in the notebook folder or change the path')
    df = pd.DataFrame()
```
